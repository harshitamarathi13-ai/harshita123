```python

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Supported target languages and their pretrained models
LANGUAGES = {
    "Hindi": "Helsinki-NLP/opus-mt-en-hi",
    "French": "Helsinki-NLP/opus-mt-en-fr",
    "Spanish": "Helsinki-NLP/opus-mt-en-es",
    "German": "Helsinki-NLP/opus-mt-en-de",
    "Italian": "Helsinki-NLP/opus-mt-en-it",
}

# Store loaded models so they do not reload for every sentence
loaded_models = {}

def translate_text(text, target_language):
    if not text.strip():
        return "Please enter some text."

    if target_language not in LANGUAGES:
        return "Unsupported target language."

    try:
        if target_language not in loaded_models:
            model_name = LANGUAGES[target_language]
            print(f"Loading {target_language} translation model...")

            tokenizer = AutoTokenizer.from_pretrained(model_name)
            model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

            loaded_models[target_language] = (tokenizer, model)

        tokenizer, model = loaded_models[target_language]

        inputs = tokenizer(
            text,
            return_tensors="pt",
            truncation=True,
            max_length=128
        )

        outputs = model.generate(
            **inputs,
            max_new_tokens=128
        )

        return tokenizer.decode(
            outputs[0],
            skip_special_tokens=True
        )

    except Exception as error:
        return f"Translation error: {error}"


# Test the translator
text = "Hello, how are you?"

for language in LANGUAGES:
    print(f"\nTarget language: {language}")
    print("English:", text)
    print("Translation:", translate_text(text, language))
```

```python

import torch
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Multilingual pretrained translation model
MODEL_NAME = "facebook/nllb-200-distilled-600M"

print("Loading translation model...")

device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
model = model.to(device)
model.eval()

print("Model loaded successfully!")

# Languages supported by the model
LANGUAGES = {
    "English": "eng_Latn",
    "Hindi": "hin_Deva",
    "Kannada": "kan_Knda",
    "Tulu": "tulu_Tulu",
    "Tamil": "tam_Taml",
    "Telugu": "tel_Telu",
    "Malayalam": "mal_Mlym",
    "Bengali": "ben_Beng",
    "Gujarati": "guj_Gujr",
    "Marathi": "mar_Deva",
    "Punjabi": "pan_Guru",
    "Urdu": "urd_Arab",
    "Odia": "ory_Orya",
    "Assamese": "asm_Beng",
    "Nepali": "npi_Deva",
    "Sanskrit": "san_Deva",
    "French": "fra_Latn",
    "Spanish": "spa_Latn",
    "German": "deu_Latn",
    "Italian": "ita_Latn",
    "Portuguese": "por_Latn",
    "Russian": "rus_Cyrl",
    "Arabic": "arb_Arab",
    "Chinese (Simplified)": "zho_Hans",
    "Chinese (Traditional)": "zho_Hant",
    "Japanese": "jpn_Jpan",
    "Korean": "kor_Hang",
    "Thai": "tha_Thai",
    "Vietnamese": "vie_Latn",
    "Indonesian": "ind_Latn",
    "Malay": "zsm_Latn",
    "Swahili": "swh_Latn",
    "Turkish": "tur_Latn",
    "Dutch": "nld_Latn",
    "Polish": "pol_Latn",
    "Ukrainian": "ukr_Cyrl",
    "Romanian": "ron_Latn",
    "Greek": "ell_Grek",
    "Hebrew": "heb_Hebr",
    "Persian": "pes_Arab",
    "Swedish": "swe_Latn",
    "Finnish": "fin_Latn",
    "Danish": "dan_Latn",
    "Norwegian": "nob_Latn",
    "Czech": "ces_Latn",
    "Hungarian": "hun_Latn",
    "Serbian": "srp_Cyrl",
    "Croatian": "hrv_Latn",
    "Bulgarian": "bul_Cyrl",
    "Filipino": "tgl_Latn",
    "Irish": "gle_Latn",
    "Welsh": "cym_Latn",
    "Yoruba": "yor_Latn",
    "Zulu": "zul_Latn",
    "Amharic": "amh_Ethi",
    "Khmer": "khm_Khmr",
    "Lao": "lao_Laoo",
    "Burmese": "mya_Mymr",
    "Sinhala": "sin_Sinh",
    "Tibetan": "bod_Tibt",
    "Kazakh": "kaz_Cyrl",
    "Uzbek": "uzn_Latn",
    "Mongolian": "khk_Cyrl",
    "Albanian": "als_Latn",
    "Armenian": "hye_Armn",
    "Georgian": "kat_Geor",
    "Icelandic": "isl_Latn",
    "Latin": "lat_Latn"
}

def translate_text(text, source_language, target_language):
    if not text or not text.strip():
        return "Please enter text to translate."

    if source_language == target_language:
        return text

    try:
        tokenizer.src_lang = LANGUAGES[source_language]

        inputs = tokenizer(
            text.strip(),
            return_tensors="pt",
            truncation=True,
            max_length=512
        ).to(device)

        target_id = tokenizer.convert_tokens_to_ids(
            LANGUAGES[target_language]
        )

        if target_id == tokenizer.unk_token_id:
            return "The selected target language is not supported by this model."

        with torch.no_grad():
            translated_tokens = model.generate(
                **inputs,
                forced_bos_token_id=target_id,
                max_new_tokens=200,
                num_beams=4
            )

        result = tokenizer.batch_decode(
            translated_tokens,
            skip_special_tokens=True
        )[0]

        return result

    except Exception as error:
        return f"Translation error: {error}"

# Text-only user interface
demo = gr.Interface(
    fn=translate_text,
    inputs=[
        gr.Textbox(
            lines=5,
            label="Enter text",
            placeholder="Type text here..."
        ),
        gr.Dropdown(
            choices=list(LANGUAGES.keys()),
            value="English",
            label="Source language"
        ),
        gr.Dropdown(
            choices=list(LANGUAGES.keys()),
            value="Hindi",
            label="Target language"
        )
    ],
    outputs=gr.Textbox(
        lines=5,
        label="Translated text"
    ),
    title="AI Language Translation App",
    description=(
        "Translate text between supported languages "
        "using a pretrained multilingual deep learning model."
    )
)

demo.launch(share=True)
```
