```python
import pandas as pd
import numpy as np
Series_A = pd.Series([10, 20, 30, 40, 50])
Series_B = pd.Series([40, 50, 60, 70, 80])
union = pd.Series(np.union1d(Series_A.values, Series_B.values))
intersect = pd.Series(np.intersect1d(Series_A.values, Series_B.values))
notcommanSeries=union[-union.isin(intersect)]
print(notcommanSeries)
print("largest in Series_A is:",Series_A.max())
print("smallest number in Series_A is:",Series_A.min())
print("the sum of Series_B is:",Series_B.sum())
print("the avarage of Series_A is:",Series_A.mean())
print("the median of Series_B is :",Series_B.median())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv('mtcars.csv')
print(cars_df)
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
print(a)
plt.xlabel('MPG')
plt.ylabel('Weight')
plt.title('Relationship between the height of the cars and mpg')
plt.show()
trans=cars_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel("transmission")
plt.title("Barplottocheckthefrequencydistributionoftransmissiontyp")
plt.show()
cars_df['mpg'].plot(kind='box')
plt.xlabel('MPG')
plt.title('Boxandwhiskerplotofmpg')
plt.show()
minvalue=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
maxvalue=np.max(cars_df['mpg'])
print('minimumvalue:',minvalue)
print('Q1:',q1)
print('median:',median)
print('Q3:',q3)
print('maxvalue:',maxvalue)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'Ename':['roshan','amar','ashwin','lohit','mohan','pramod'],
    'Department':['cs','cs','ec','ec','cs','ec'],
    'Experience':[10,20,15,14,9,8],
    'Salary':[45000,50000,30000,25000,10000,40000]
    }
df=pd.DataFrame(data)
print(df)
table = pd.pivot_table(df, values='Salary', index=['Ename'], columns=['Department'], aggfunc=np.mean)
print(table)
table = pd.pivot_table(df, values='Salary', index=['Ename'], aggfunc=['sum', 'mean'])
print(table)
```
